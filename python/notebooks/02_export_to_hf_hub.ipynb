````python
import os
from orobot import OrobotClient

client = OrobotClient(api_key=os.environ.get('OROBOT_API_KEY', 'ork_live_...'))
````

## Export

`export_lerobot_dataset` walks every episode under `data_dir`, maps each frame's
joints to the LeRobotDataset observation/action columns, encodes camera frames to
MP4, writes Parquet shards, and (optionally) pushes to the Hub.

````python
client.export_lerobot_dataset(
    './data/',
    repo_id='myuser/my-robot-data',
    push_to_hub=True,  # requires `huggingface-cli login`
)
````

## Verify on the Hub

Once implemented, load it straight back with LeRobot to confirm the round-trip:
